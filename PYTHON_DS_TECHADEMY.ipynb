```python
# single inheritance --> one child class inherites from one
#   parent class is called single inheritance
```

```python
class Mobile:
  def Brand(self):
    print("Brand: Samsung")
class Smartphone(Mobile):
  def Feature(self):
    print("Feature: Face Unlock")

c=Smartphone()
c.Feature()
c.Brand()
```

```python
##Multilevel inheritance-->Grand parent---Parent---Child
# A child class inherites from a parent class and that parent class
# is also called inherited from a super parent class
```

```python
class GrandParent:
  def g_land(self):
    print("This is Grand parent land")

class Parent(GrandParent):
  def p_land(self):
    print("This is parent land")

class Child(Parent):
  def c_land(self):
    print("This is child land")

m=Child()
m.g_land()
m.p_land()
m.c_land()
```

```python
##  Multiple Inheritance--> Multiple parents and single class


class Father:
  def f_land(self):
    print("This is father land")
class Mother:
  def m_land(self):
    print("This is mother land")
class Child(Father,Mother):
  def c_land(self):
    print("This is child land")

m=Child()
m.f_land()
m.m_land()
m.c_land()
```

```python

class GrandParent:
  def g_land(self):
    print("This is Grand parent land")

class Parent():
  def p_land(self):
    print("This is parent land")

class Child(GrandParent,Parent):
  def c_land(self):
    print("This is child land")

m=Child()
m.g_land()
m.p_land()
m.c_land()
```

```python
#Hierarchical--->  one parent many child


class Parent:
  def p_land(self):
    print("This is parent land")

class Child1(Parent):
  def c1_land(self):
    print("This is child1 land")

class Child2(Parent):
  def c2_land(self):
    print("This is child2 land")

h=Child1()
h.p_land()
h.c1_land()

print("")

g=Child2()
g.p_land()
g.c2_land()
```

```python

class Animal:
  def __init__(self): # constructor(automaticalley called)
    print("animala eats food")
  def bark(self):
    print("Dog Barks")

class Dog(Animal):
  def __init__(self):#constructor
    super().__init__()#here we are calling parent class consturctor
    print("This is spl function of dog class")
    super().bark()
  def food(self):
    print("This is spl function of dog class")



d=Dog()
```
